```python
df2 = df2.rename(columns={
    "Invoice": "InvoiceNo",
    "Price": "UnitPrice",
    "Customer ID": "CustomerID",
})
print(df1.columns.tolist())
print(df2.columns.tolist())
```

```python
print("File 1:", df1["InvoiceDate"].min(), "->", df1["InvoiceDate"].max())
print("File 2:", df2["InvoiceDate"].min(), "->", df2["InvoiceDate"].max())
```

```python
keys = ["InvoiceNo", "StockCode", "InvoiceDate", "Quantity", "UnitPrice"]

a = df1.copy()
b = df2[df2["InvoiceDate"] >= df1["InvoiceDate"].min()].copy()

for d in (a, b):
    d["InvoiceNo"] = d["InvoiceNo"].astype(str)
    d["StockCode"] = d["StockCode"].astype(str)
    d["_n"] = d.groupby(keys).cumcount()   # đánh số các dòng giống hệt nhau

m = a.merge(b[keys + ["_n"]], on=keys + ["_n"], how="left", indicator=True)
print(len(a), len(b))
print(m["_merge"].value_counts())
```
